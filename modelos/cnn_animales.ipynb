```python
import tensorflow as tf

# Cargar modelo original
model = tf.keras.models.load_model("cnn_animales.h5")

# Convertir a TFLite
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

# Guardar archivo
with open("cnn_animales.tflite", "wb") as f:
    f.write(tflite_model)

print("Conversión completada: cnn_animales.tflite")
```
